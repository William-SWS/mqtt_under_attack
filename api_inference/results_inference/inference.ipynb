```python
import pandas as pd 
from pathlib import Path

base = Path("api_inference/results_inference")
candidates = [
    base / "iinference.csv",
    base / "inference.csv",
]
csv_path = None

if base.exists():
    for p in candidates:
        if p.exists():
            csv_path = p
            break
    if csv_path is None:
        files = sorted(base.glob("*.csv"))
        if files:
            csv_path = files[0]
else:
    matches = list(Path(".").rglob("*inference*.csv"))
    if matches:
        csv_path = matches[0]

if csv_path is None:
    raise FileNotFoundError(
        "CSV not found. Checked: 'api_inference/results_inference/iinference.csv', "
        "'api_inference/results_inference/inference.csv', directory CSVs, and recursive search."
    )

df = pd.read_csv(csv_path)

df.sort_values("inference_time_ms", ascending=True).head(20)
```
